```python

data = [
    {"name": "John Doe", "age": 30, "city": "New York"},
    {"name": "Jane Smith", "age": 25, "city": "Los Angeles"},
    {"name": "Bob Johnson", "age": 35, "city": "Chicago"}
]
```

```python
for rows in data:
    print(rows)
```

```python
import sqlite3

conn = sqlite3.connect('persoon.db')
cur  = conn.cursor()
cur.execute('create table if not exists persoon(persoon_id int, name string, age int, city string)')

conn.close()
```

```python
import sqlite3
import pygrametl
from pygrametl.tables import Dimension

pgconn = pygrametl.ConnectionWrapper(sqlite3.connect('persoon.db'))

persoonDimension = Dimension(name='persoon',
                             key='persoon_id',
                             attributes=['name', 'age', 'city'],
                             lookupatts=['name'],
                             targetconnection=pgconn)
```

```python
for row in data:
    persoonDimension.insert(row)

pgconn.commit()
pgconn.close()
```
